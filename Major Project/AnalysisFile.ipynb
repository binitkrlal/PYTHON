```python
import numpy as np
import pandas as pd
```

```python
#Loading the file into a pandas dataframe using separator as None and engine as python to avoid any parsing errors
df = pd.read_csv("mymoviedb.csv", sep=None, engine="python")
df.head()
```

```python
#Finding the data types of each columns
df.info()
```

```python
#Finding the number of duplicate rows in the dataset
df.duplicated().sum()
```

```python
#Describing the dataset to get a statistical summary of the numerical columns
df.describe()
```

```python
#Converting the Vote_Count and Vote_Average columns to numeric data types, coercing errors to NaN
df["Vote_Count"] = pd.to_numeric(df["Vote_Count"], errors="coerce")
df["Vote_Average"] = pd.to_numeric(df["Vote_Average"], errors="coerce")

df.describe()
```

```python
#vote_count and vote_average columns are converted and confirmed the changes
df.info()
```

```python
#converting the Release_Date column to datetime format, coercing errors=lets the conversion continue, but turns invalid date values into NaT
df["Release_Date"] = pd.to_datetime(df["Release_Date"], errors="coerce")
print(df["Release_Date"].dtypes)
```

```python
# check which rows couldn’t be converted:
df.loc[df["Release_Date"].isna(), ["Title", "Release_Date"]].head()
```

```python
# converting the date into int and checking its types and keeping only released year in the Release_Date column
df["Release_Date"] = df["Release_Date"].dt.year
df["Release_Date"].dtypes
```

```python
# converting the Release_Date column to Int64 data type to handle missing values
df["Release_Date"] = df["Release_Date"].astype("Int64")

# checking the result from int conversion
df.head()
```

```python
# Dropping the columns that are not needed for analysis
cols = ["Overview", "Original_Language", "Poster_Url"]
df.drop(columns=cols, inplace=True, errors="ignore")

df.head()
```

```python
# Creating the user defined function to labels the vote_average column into 5 categories based on the given conditions
# It is aa user defined function.
def categorize_col(df, col, labels):
    edges = [df[col].describe()["min"],
             df[col].describe()["25%"],
             df[col].describe()["50%"],
             df[col].describe()["75%"],
             df[col].describe()["max"]]

    df[col] = pd.cut(df[col], edges, labels=labels, duplicates="drop")

    return df
```

```python
#Categorizing Vote Column into 4 categories based on the given conditions and storing the labels in a list
labels = ["not_popular", "below_avg", "average", "popular"]

categorize_col(df, "Vote_Average", labels)

df["Vote_Average"].unique()
```

```python
df.head()
```

```python
#Finding the value counts of the Vote_Average column to see how many movies fall into each category
df['Vote_Average'].value_counts()
```

```python
#Dropping the duplicate  and nan values
df.dropna(inplace=True)
df.isna().sum()
```

```python
#Splitting our genre to explode our dataframe more to see how many fall in each genre.
df['Genre'] = df['Genre'].str.split(', ')

df = df.explode('Genre').reset_index(drop=True)

df.head()
```

```python
#Casting coloumn into Category
df["Genre"] = df["Genre"].astype("category")
df["Genre"].dtypes
```
